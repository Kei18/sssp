# SSSP demo, all scenarios

```julia
import Random: seed!
using JLD2
import Printf: @sprintf, @printf
using MRMP
```

```julia
benchmark_dir = "../../data/benchmark/ijcai23/main"
```

# point2d

```julia
scenario ="point2d"
benchmark_file = joinpath(benchmark_dir, "$(scenario).jld2")
I = JLD2.load(benchmark_file, "instances")
ins = I[5]
config_init, config_goal, obstacles, ins_params... = ins
connect = gen_connect(config_init[1], obstacles, ins_params...)
collide = gen_collide(config_init[1], ins_params...)
check_goal = gen_check_goal(config_goal)
planner_args = [config_init, config_goal, connect, collide, check_goal]
MRMP.plot_instance!(ins...)
```

```julia
seed!(0)
@time solution, roadmaps = MRMP.SSSP(
    planner_args...; 
    TIME_LIMIT = 10,
    VERBOSE = 2,
)
MRMP.validate(config_init, connect, collide, check_goal, solution)
@time res = smoothing(solution, connect, collide)
nothing
```

```julia
plot_res!(config_init, config_goal, obstacles, ins_params...; solution=res[2], filename = "./fig/point2d.pdf")
```

```julia
plot_anim!(config_init, config_goal, obstacles, ins_params...; solution=res[2], interpolate_depth=3, filename = "./fig/point2d.gif", fps=30, VERBOSE=1)
```

# point2d, many agents

```julia
benchmark_file = "../../data/benchmark/ijcai23/scalability/point2d_30.jld2"
I = JLD2.load(benchmark_file, "instances")
ins = I[3]
config_init, config_goal, obstacles, ins_params... = ins
connect = gen_connect(config_init[1], obstacles, ins_params...)
collide = gen_collide(config_init[1], ins_params...)
check_goal = gen_check_goal(config_goal)
planner_args = [config_init, config_goal, connect, collide, check_goal]
MRMP.plot_instance!(ins...)
```

```julia
seed!(1)
@time solution, roadmaps = MRMP.Solvers.SSSP(planner_args...;
    VERBOSE = 1,
    TIME_LIMIT = 300,
    steering_depth = 2,
    num_vertex_expansion = 10,
    init_min_dist_thread = 0.05,
    epsilon = 0.2,
)
MRMP.validate(config_init, connect, collide, check_goal, solution)
@time res = smoothing(solution, connect, collide; VERBOSE=1)
nothing
```

```julia
plot_res!(config_init, config_goal, obstacles, ins_params...; solution=res[2], filename = "./fig/point2d_many.pdf")
```

```julia
plot_anim!(config_init, config_goal, obstacles, ins_params...; solution=res[2], interpolate_depth=3, fps=30, VERBOSE=1, filename = "./fig/point2d_many.gif")
```
